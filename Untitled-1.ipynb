## 3. Resize and Save Images

Resize all images to a uniform size to speed up training. This preprocessing step is done once to avoid repeated resizing during each epoch.

```python
start_time = time.time()

# Image dimensions
img_height = 300
img_width = 300

# Remove existing directory and create fresh one
if os.path.isdir(save_path):
    shutil.rmtree(save_path)
os.makedirs(save_path, exist_ok=True)

# Process each class directory
classlist = sorted(os.listdir(sdir))
total_images = 0
error_count = 0

print("Starting image resizing process...")
print("="*60)

for klass in classlist:
    classpath = os.path.join(sdir, klass)
    
    # Skip if not a directory
    if not os.path.isdir(classpath):
        continue
    
    print(f'Processing class: {klass}')
    
    # Create class directory in save path
    class_dest_path = os.path.join(save_path, klass)
    os.makedirs(class_dest_path, exist_ok=True)
    
    # Process each image in the class
    flist = os.listdir(classpath)
    class_image_count = 0
    
    for f in flist:
        fpath = os.path.join(classpath, f)
        f_dest_path = os.path.join(class_dest_path, f)
        
        try:
            # Read, resize, and save image
            img = cv2.imread(fpath)
            if img is None:
                raise ValueError("Invalid image file")
            img = cv2.resize(img, (img_width, img_height))
            cv2.imwrite(f_dest_path, img)
            class_image_count += 1
            total_images += 1
        except Exception as e:
            print(f'  ⚠️ Error processing {fpath}: {str(e)}')
            error_count += 1
    
    print(f'  ✅ Processed {class_image_count} images')

# Calculate and display elapsed time
tr_duration = time.time() - start_time
hours = tr_duration // 3600
minutes = (tr_duration - (hours * 3600)) // 60
seconds = tr_duration - ((hours * 3600) + (minutes * 60))

print("="*60)
print(f'Total images processed: {total_images}')
print(f'Errors encountered: {error_count}')
print(f'Resizing elapsed time: {int(hours)} hours, {minutes:4.1f} minutes, {seconds:4.2f} seconds')
print("="*60)
```

## 4. Create DataFrames from Image Paths

Build a structured DataFrame containing image paths and labels, then split into training, validation, and test sets.

```python
def make_df(save_path):
    """
    Create a DataFrame from image files and split into train, validation, and test sets.
    
    Args:
        save_path: Path to directory containing resized images
        
    Returns:
        train_df, test_df, valid_df, classes
    """
    classlist = os.listdir(save_path)
    filepaths = []
    labels = []
    
    # Collect all file paths and labels
    for klass in classlist:
        classpath = os.path.join(save_path, klass)
        flist = os.listdir(classpath)
        for f in flist:
            fpath = os.path.join(classpath, f)
            filepaths.append(fpath)
            labels.append(klass)
    
    # Create DataFrame
    Fseries = pd.Series(filepaths, name='filepaths')
    Lseries = pd.Series(labels, name='labels')
    df = pd.concat([Fseries, Lseries], axis=1)
    
    # Split data: 80% train, 10% validation, 10% test
    train_df, dummy_df = train_test_split(
        df, train_size=0.8, shuffle=True, random_state=123, stratify=df['labels']
    )
    valid_df, test_df = train_test_split(
        dummy_df, train_size=0.5, shuffle=True, random_state=123, stratify=dummy_df['labels']
    )
    
    print('='*70)
    print(f'Train set length: {len(train_df)}')
    print(f'Validation set length: {len(valid_df)}')
    print(f'Test set length: {len(test_df)}')
    print('='*70)
    
    # Get class information
    classes = sorted(list(train_df['labels'].unique()))
    class_count = len(classes)
    print(f'\nNumber of classes in the dataset: {class_count}\n')
    
    # Display class distribution
    groups = train_df.groupby('labels')
    print('{0:^30s} {1:^13s}'.format('CLASS', 'IMAGE COUNT'))
    print('-'*45)
    
    countlist = []
    classlist = []
    
    for label in sorted(list(train_df['labels'].unique())):
        group = groups.get_group(label)
        count = len(group)
        countlist.append(count)
        classlist.append(label)
        print('{0:^30s} {1:^13s}'.format(label, str(count)))
    
    # Find min and max classes
    max_value = np.max(countlist)
    max_index = countlist.index(max_value)
    max_class = classlist[max_index]
    min_value = np.min(countlist)
    min_index = countlist.index(min_value)
    min_class = classlist[min_index]
    
    print('-'*45)
    print(f'{max_class} has the most images: {max_value}')
    print(f'{min_class} has the least images: {min_value}')
    
    # Calculate average image dimensions
    ht = 0
    wt = 0
    train_df_sample = train_df.sample(n=min(100, len(train_df)), random_state=123, axis=0)
    
    for i in range(len(train_df_sample)):
        fpath = train_df_sample['filepaths'].iloc[i]
        img = plt.imread(fpath)
        shape = img.shape
        ht += shape[0]
        wt += shape[1]
    
    avg_ht = ht // len(train_df_sample)
    avg_wt = wt // len(train_df_sample)
    aspect_ratio = ht / wt
    
    print(f'\nAverage height: {avg_ht}')
    print(f'Average width: {avg_wt}')
    print(f'Aspect ratio: {aspect_ratio:.2f}')
    print('='*70)
    
    return train_df, test_df, valid_df, classes

# Create DataFrames
train_df, test_df, valid_df, classes = make_df(save_path)
class_count = len(classes)
```

<a id="balance"></a>
## 5. Balance Dataset with Data Augmentation

Create augmented images for underrepresented classes to balance the dataset and improve model generalization.

```python
def balance(df, n, working_dir, img_size):
    """
    Balance the dataset by creating augmented images for classes with fewer samples.
    
    Args:
        df: DataFrame with image paths and labels
        n: Target number of samples per class
        working_dir: Directory to store augmented images
        img_size: Tuple of (height, width) for augmented images
        
    Returns:
        Balanced DataFrame
    """
    df = df.copy()
    print(f'Initial dataframe length: {len(df)}')
    
    # Create augmentation directory
    aug_dir = os.path.join(working_dir, 'aug')
    if os.path.isdir(aug_dir):
        shutil.rmtree(aug_dir)
    os.mkdir(aug_dir)
    
    # Create class directories
    for label in df['labels'].unique():
        dir_path = os.path.join(aug_dir, label)
        os.mkdir(dir_path)
    
    # Configure augmentation generator
    gen = ImageDataGenerator(
        horizontal_flip=True,
        rotation_range=20,
        width_shift_range=0.2,
        height_shift_range=0.2,
        zoom_range=0.2
    )
    
    total = 0
    groups = df.groupby('labels')
    
    print('\nCreating augmented images...')
    print('='*60)
    
    for label in df['labels'].unique():
        group = groups.get_group(label)
        sample_count = len(group)
        
        if sample_count < n:
            aug_img_count = 0
            delta = n - sample_count
            target_dir = os.path.join(aug_dir, label)
            
            msg = f'Class {label:^30s}: creating {delta:^5d} augmented images'
            print(msg)
            
            aug_gen = gen.flow_from_dataframe(
                group,
                x_col='filepaths',
                y_col=None,
                target_size=img_size,
                class_mode=None,
                batch_size=1,
                shuffle=False,
                save_to_dir=target_dir,
                save_prefix='aug-',
                color_mode='rgb',
                save_format='jpg'
            )
            
            while aug_img_count < delta:
                images = next(aug_gen)
                aug_img_count += len(images)
            
            total += aug_img_count
    
    print('='*60)
    print(f'Total augmented images created: {total}')
    
    # Create DataFrame from augmented images
    aug_fpaths = []
    aug_labels = []
    classlist = os.listdir(aug_dir)
    
    for klass in classlist:
        classpath = os.path.join(aug_dir, klass)
        flist = os.listdir(classpath)
        for f in flist:
            fpath = os.path.join(classpath, f)
            aug_fpaths.append(fpath)
            aug_labels.append(klass)
    
    Fseries = pd.Series(aug_fpaths, name='filepaths')
    Lseries = pd.Series(aug_labels, name='labels')
    aug_df = pd.concat([Fseries, Lseries], axis=1)
    
    # Merge with original DataFrame
    df = pd.concat([df, aug_df], axis=0).reset_index(drop=True)
    print(f'Final dataframe length: {len(df)}')
    print('='*60)
    
    return df

# Balance the training set
n = 200  # Target number of samples per class
img_size = (300, 300)
train_df = balance(train_df, n, working_dir, img_size)
```

## 8. Build EfficientNetB3 Model with Mixed Precision

Create a transfer learning model using EfficientNetB3 as the base. **Critical fix:** The output layer must use `dtype='float32'` for mixed precision compatibility.

```python
img_shape = (img_size[0], img_size[1], 3)
model_name = 'EfficientNetB3'

print('Building model...')
print('='*70)

# Create base model
base_model = tf.keras.applications.efficientnet.EfficientNetB3(
    include_top=False,
    weights="imagenet",
    input_shape=img_shape,
    pooling='max'
)

# Make base model trainable for better results
base_model.trainable = True

# Build custom top layers
x = base_model.output
x = BatchNormalization(axis=-1, momentum=0.99, epsilon=0.001)(x)
x = Dense(
    256,
    kernel_regularizer=regularizers.l2(0.016),
    activity_regularizer=regularizers.l1(0.006),
    bias_regularizer=regularizers.l1(0.006),
    activation='relu'
)(x)
x = Dropout(rate=0.4, seed=123)(x)

# CRITICAL FIX: Output layer MUST be float32 for mixed precision
output = Dense(class_count, activation='softmax', dtype='float32')(x)

# Create model
model = Model(inputs=base_model.input, outputs=output)

# Compile model
lr = 0.001
model.compile(
    Adamax(learning_rate=lr),
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Verify the fix
print(f"✅ Model output dtype: {model.output.dtype}")
print(f"✅ Base model trainable: {base_model.trainable}")
print(f"✅ Total parameters: {model.count_params():,}")
print(f"✅ Model compiled successfully!")
print('='*70)

# Display model summary
model.summary()
```

## 9. Define Custom Training Callback

Implement a custom callback for interactive training control with the ability to adjust learning rate or halt training.

```python
class LR_ASK(keras.callbacks.Callback):
    """
    Custom callback for interactive training control.
    
    Features:
    - Pause training at specified epochs
    - Adjust learning rate interactively
    - Save best weights based on validation loss
    - Track training duration
    """
    
    def __init__(self, model, epochs, ask_epoch):
        super(LR_ASK, self).__init__()
        self.user_model = model
        self.ask_epoch = ask_epoch
        self.epochs = epochs
        self.ask = True
        self.lowest_vloss = np.inf
        self.best_weights = self.user_model.get_weights()
        self.best_epoch = 1
    
    def on_train_begin(self, logs=None):
        if self.ask_epoch == 0:
            print('ask_epoch set to 0, changing to 1', flush=True)
            self.ask_epoch = 1
        
        if self.ask_epoch >= self.epochs:
            print(f'ask_epoch >= epochs, will train for {self.epochs} epochs', flush=True)
            self.ask = False
        
        if self.epochs == 1:
            self.ask = False
        else:
            print(f'Training will proceed until epoch {self.ask_epoch}')
            print('You will be asked to enter H to halt or an integer for more epochs')
        
        self.start_time = time.time()
    
    def on_train_end(self, logs=None):
        print(f'\nLoading model with weights from epoch {self.best_epoch}')
        self.user_model.set_weights(self.best_weights)
        
        tr_duration = time.time() - self.start_time
        hours = tr_duration // 3600
        minutes = (tr_duration - (hours * 3600)) // 60
        seconds = tr_duration - ((hours * 3600) + (minutes * 60))
        
        msg = f'Training elapsed time: {int(hours)} hours, {minutes:4.1f} minutes, {seconds:4.2f} seconds'
        print(msg, flush=True)
    
    def on_epoch_end(self, epoch, logs=None):
        v_loss = logs.get('val_loss')
        
        if v_loss < self.lowest_vloss:
            self.lowest_vloss = v_loss
            self.best_weights = self.user_model.get_weights()
            self.best_epoch = epoch + 1
            print(f'\n✅ Validation loss {v_loss:.4f} is new best - saving weights from epoch {epoch + 1}')
        else:
            print(f'\nValidation loss {v_loss:.4f} above best {self.lowest_vloss:.4f} - keeping epoch {self.best_epoch} weights')
        
        if self.ask and epoch + 1 == self.ask_epoch:
            print('\nEnter H to halt training or an integer for additional epochs:')
            ans = input()
            
            if ans in ['H', 'h', '0']:
                print(f'Training halted at epoch {epoch+1}', flush=True)
                self.model.stop_training = True
            else:
                self.ask_epoch += int(ans)
                if self.ask_epoch > self.epochs:
                    print(f'Cannot train beyond {self.epochs} epochs', flush=True)
                else:
                    print(f'Training continues to epoch {self.ask_epoch}', flush=True)
                    
                    # Fixed: Use learning_rate instead of lr
                    lr = float(tf.keras.backend.get_value(self.user_model.optimizer.learning_rate))
                    print(f'Current LR: {lr:.5f} - press Enter to keep or enter new LR:')
                    ans = input(' ')
                    
                    if ans == '':
                        print(f'Keeping LR of {lr:.5f}')
                    else:
                        new_lr = float(ans)
                        tf.keras.backend.set_value(self.user_model.optimizer.learning_rate, new_lr)
                        print(f'LR changed to {new_lr}')
```

## 12. Plot Training History

Visualize training and validation metrics over epochs to analyze model performance.

```python
def tr_plot(tr_data, start_epoch=0):
    """
    Plot training and validation loss and accuracy.
    
    Args:
        tr_data: History object from model.fit()
        start_epoch: Starting epoch number for plotting
    """
    tacc = tr_data.history['accuracy']
    tloss = tr_data.history['loss']
    vacc = tr_data.history['val_accuracy']
    vloss = tr_data.history['val_loss']
    
    Epoch_count = len(tacc) + start_epoch
    Epochs = list(range(start_epoch + 1, Epoch_count + 1))
    
    index_loss = np.argmin(vloss)
    val_lowest = vloss[index_loss]
    index_acc = np.argmax(vacc)
    acc_highest = vacc[index_acc]
    
    plt.style.use('fivethirtyeight')
    sc_label = f'Best epoch = {index_loss + 1 + start_epoch}'
    vc_label = f'Best epoch = {index_acc + 1 + start_epoch}'
    
    fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(20, 8))
    
    # Loss plot
    axes[0].plot(Epochs, tloss, 'r', label='Training loss')
    axes[0].plot(Epochs, vloss, 'g', label='Validation loss')
    axes[0].scatter(
        index_loss + 1 + start_epoch,
        val_lowest,
        s=150,
        c='blue',
        label=sc_label
    )
    axes[0].set_title('Training and Validation Loss', fontsize=16)
    axes[0].set_xlabel('Epochs', fontsize=14)
    axes[0].set_ylabel('Loss', fontsize=14)
    axes[0].legend()
    axes[0].grid(True)
    
    # Accuracy plot
    axes[1].plot(Epochs, tacc, 'r', label='Training Accuracy')
    axes[1].plot(Epochs, vacc, 'g', label='Validation Accuracy')
    axes[1].scatter(
        index_acc + 1 + start_epoch,
        acc_highest,
        s=150,
        c='blue',
        label=vc_label
    )
    axes[1].set_title('Training and Validation Accuracy', fontsize=16)
    axes[1].set_xlabel('Epochs', fontsize=14)
    axes[1].set_ylabel('Accuracy', fontsize=14)
    axes[1].legend()
    axes[1].grid(True)
    
    plt.tight_layout()
    plt.show()
    
    # Print summary statistics
    print('='*70)
    print(f'Best validation loss: {val_lowest:.4f} at epoch {index_loss + 1 + start_epoch}')
    print(f'Best validation accuracy: {acc_highest:.4f} at epoch {index_acc + 1 + start_epoch}')
    print(f'Final training loss: {tloss[-1]:.4f}')
    print(f'Final training accuracy: {tacc[-1]:.4f}')
    print('='*70)

tr_plot(history, 0)
```

## 13. Evaluate Model Performance

Generate predictions on the test set and calculate overall accuracy.

```python
def evaluate_model(model, test_gen):
    """
    Evaluate model performance on test set.
    
    Args:
        model: Trained Keras model
        test_gen: Test data generator
        
    Returns:
        y_pred, y_true, errors, tests
    """
    print('Evaluating model on test set...')
    print('='*70)
    
    y_pred = []
    y_true = test_gen.labels
    classes = list(test_gen.class_indices.keys())
    errors = 0
    error_list = []
    
    # Generate predictions
    preds = model.predict(test_gen, verbose=1)
    tests = len(preds)
    
    for i, p in enumerate(preds):
        pred_index = np.argmax(p)
        true_index = test_gen.labels[i]
        
        if pred_index != true_index:
            errors += 1
            file = test_gen.filenames[i]
            true_class = classes[true_index]
            pred_class = classes[pred_index]
            confidence = p[pred_index] * 100
            error_list.append({
                'file': file,
                'true': true_class,
                'predicted': pred_class,
                'confidence': confidence
            })
        
        y_pred.append(pred_index)
    
    acc = (1 - errors / tests) * 100
    
    print('='*70)
    print(f'Test Results:')
    print(f'  Total samples: {tests}')
    print(f'  Errors: {errors}')
    print(f'  Accuracy: {acc:.2f}%')
    print('='*70)
    
    # Show first 10 errors
    if error_list:
        print('\nFirst 10 misclassifications:')
        print('-'*70)
        for i, err in enumerate(error_list[:10]):
            print(f"{i+1}. {err['file']}")
            print(f"   True: {err['true']} | Predicted: {err['predicted']} ({err['confidence']:.1f}%)")
        print('-'*70)
    
    return np.array(y_pred), np.array(y_true), errors, tests

y_pred, y_true, errors, tests = evaluate_model(model, test_gen)
```

## 14. Generate Predictions and Confusion Matrix

Create detailed performance metrics including confusion matrix and classification report.

```python
def plot_confusion_matrix(y_true, y_pred, classes):
    """
    Plot confusion matrix heatmap and display classification report.
    
    Args:
        y_true: True labels
        y_pred: Predicted labels
        classes: List of class names
    """
    class_count = len(classes)
    
    if class_count <= 30:
        # Create confusion matrix
        cm = confusion_matrix(y_true, y_pred)
        
        # Plot heatmap
        plt.figure(figsize=(max(12, class_count), max(8, class_count * 0.7)))
        sns.heatmap(
            cm,
            annot=True,
            fmt='g',
            cmap='Blues',
            cbar=True,
            xticklabels=classes,
            yticklabels=classes
        )
        plt.xlabel('Predicted', fontsize=14)
        plt.ylabel('Actual', fontsize=14)
        plt.title('Confusion Matrix', fontsize=16)
        plt.xticks(rotation=90)
        plt.yticks(rotation=0)
        plt.tight_layout()
        plt.show()
    else:
        print('Too many classes for confusion matrix visualization')
    
    # Classification report
    print('\n' + '='*70)
    print('Classification Report:')
    print('='*70)
    clr = classification_report(y_true, y_pred, target_names=classes, digits=4)
    print(clr)
    print('='*70)

plot_confusion_matrix(y_true, y_pred, classes)
```
